# Control Loops

```python
#IF-ELIF-ELSE

secret = 1.2

if secret%2==0:
    print("even")
elif secret%2==1:
    print("odd")
elif secret == 0:
    print("Neither even not odd")
else:
    print("Not a whole number")

# FOR LOOP

for i in range(10):
    print(i)

# WHILE LOOP
guess = 0
while guess!= secret:
    guess += 0.1
    print(guess)
```

# Functions

```python
# A simple custom function for a mathematical formula.

def compound_interest(principal, rate, years, times_per_year=1):
    """Return the final amount using A = P * (1 + r/n) ** (n * t).

    principal      : starting amount (P)
    rate           : annual interest rate as a decimal, e.g. 0.05 for 5% (r)
    years          : number of years (t)
    times_per_year : compounding frequency (n)
    """
    return principal * (1 + rate / times_per_year) ** (times_per_year * years)

print(compound_interest(100, 0.7, 3))
```

## Wrapper vs Function

Not all functions are wrappers. While wrappers are functions, a function only becomes a "wrapper" if its primary intent is to act as a middleman for another function or system call.
The distinction lies entirely in what the function does with its workload.

```python
def calculate_factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

def safe_factorial(n):
    if n < 0:
        raise ValueError("Number must be positive")
    return calculate_factorial(n)  # Calling the underlying function
```

### What are *args and **kwargs?

They allow a function to accept a variable number of arguments.

### Why *args and **kwargs Are Needed:
They are essential when writing a wrapper or decorator that needs to accept and forward any arbitrary arguments to an underlying function, without needing to know that function's specific signature in advance.

```python
def my_func(*args):
    print(args)

my_func(1, 2, 3)
# (1, 2, 3)
```

```python
def my_func(**kwargs):
    print(kwargs)

my_func(name="Sam", age=20)
# {'name': 'Sam', 'age': 20}
```

### Wrapper vs Decorator

People often use decorator and wrapper interchangeably, but they refer to two distinct parts of the same mechanism:

The Decorator is the outer function (the factory). Its job is to take a target function as input, set up the modification, and return the new function.

The Wrapper is the inner function. Its job is to intercept the function call, handle arguments (*args, **kwargs), execute extra logic, and invoke the original function.

```python
def my_decorator(func):
    def wrapper():
        print("Before execution")
        func()
        print("After execution")
    return wrapper

def say_hello():
    print("Hello!")

# --- Method 1: Without @ (Manual Wrapping) ---
custom_func = my_decorator(say_hello)
custom_func()
print("="*20)
# --- Method 2: With @ (Decorator Syntax) ---
@my_decorator
def say_hello_decorated():
    print("Hello!")

say_hello_decorated()
```

# Classes

```python
# Classes, inheritance and basic magic (dunder) methods.

class Animal:
    """Base (parent) class."""

    def __init__(self, name, age):       # magic method: constructor
        self.name = name
        self.age = age

    def speak(self):
        return f"{self.name} makes a sound"

    def __str__(self):                   # magic method: print(obj)
        return f"{self.__class__.__name__}(name={self.name}, age={self.age})"

    def __repr__(self):                  # magic method: debugging representation
        return f"{self.__class__.__name__}({self.name!r}, {self.age!r})"

    def __eq__(self, other):             # magic method: ==
        return self.name == other.name and self.age == other.age

    def __len__(self):                   # magic method: len(obj)
        return len(self.name)

    def __add__(self, other):            # magic method: +
        return f"{self.name} & {other.name}"


class Dog(Animal):
    """Child class: inherits everything from Animal."""

    def __init__(self, name, age, breed):
        super().__init__(name, age)      # call the parent's constructor
        self.breed = breed

    def speak(self):                     # method overriding
        return f"{self.name} says Woof!"


class Cat(Animal):
    """Another child class overriding speak()."""

    def speak(self):
        return f"{self.name} says Meow!"


if __name__ == "__main__":
    d = Dog("Bruno", 3, "Labrador")
    c = Cat("Misty", 2)

    print(d)                             # __str__
    print(repr(c))                       # __repr__
    print(d.speak())                     # overridden method
    print(c.speak())
    print(len(d))                        # __len__
    print(d + c)                         # __add__
    print(d == Dog("Bruno", 3, "Lab"))   # __eq__
    print(isinstance(d, Animal))         # True: a Dog is an Animal
```

Magic methods are special methods whose names start and end with double underscores, like __init__() and __str__(). Because of the double underscores, they are also called dunder methods (short for "double underscore").

You do not call magic methods directly. Python calls them for you, automatically, in certain situations - like when an object is created, printed, or compared to another object.

# Modules and Packages

A module is a single Python file (.py) containing code like functions, classes, and variables, while a package is a directory that holds multiple modules (and possibly sub-packages) along with an __init__.py file.

## Modules

User-Created Module: A .py file you write yourself in your project (e.g., my_tools.py) to organize your own reusable code. <br> You import it using import my_tools.

Preexisting Module:<br>
	• Built-in modules: Written in Python or C and shipped directly with the Python standard library (e.g., math, random, os, sys).<br>
	• Read the Python Modules Documentation for details on inspecting and importing built-in files


## Packages

• User-Created Package: A custom folder you create containing multiple related .py modules and an __init__.py file to structure a larger application.<br>


• Preexisting Package: <br>
	• Third-party external packages: Created by the community and installed via package managers like pip (e.g., numpy, pandas, requests, django). 

# Homework: How to publish a package?

### What a Python Package Should Contain and Why
A standard, production-ready Python package requires specific files to be recognized, tested, built, and properly hosted on the Python Package Index (PyPI).
- pyproject.toml: The modern configuration file (PEP 621). It defines your package's metadata (name, version, dependencies, author) and specifies the build system backend (e.g., setuptools or hatchling). Why: PyPI and build tools rely on this file to understand what your package is and how to compile it.
- README.md: A markdown file detailing what your package does, installation instructions, and basic usage examples. Why: This content automatically becomes the homepage of your project on PyPI, helping users decide whether to use it.
- LICENSE: A legal document (e.g., MIT, Apache 2.0) defining how others can use, modify, and distribute your code. Why: Without a license, standard copyright laws apply, meaning technically no one has the legal right to use your code.
- src/ or package directory: The actual folder containing your Python modules and an __init__.py file. Why: This holds the raw logic that users will import when they run import your_package.
- tests/ folder: Contains unit tests (using pytest or unittest). Why: Ensures your code functions correctly across updates and builds user trust.
